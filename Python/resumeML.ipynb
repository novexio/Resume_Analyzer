```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
import joblib

df = pd.read_csv("resume-dataset.csv", sep=",")

features = ["text"]
target = "role"
# print(df.columns.tolist())
# print(df.head())
# print(df.shape)

X = df[features]
y = df[target]

# X split data set

X_train,X_test,y_train,y_test = train_test_split(
    X, y, random_state = 40, test_size = 0.2  
)
```

```python
# Converting text to numbers using TfidfVectorizer
vectorizer = TfidfVectorizer()
X_train_vectorized = vectorizer.fit_transform(X_train['text'])
X_test_vectorized = vectorizer.transform(X_test['text'])

# Model training
model = MultinomialNB()
model.fit(X_train_vectorized, y_train)
joblib.dump(vectorizer, "vectorizer.pkl")
joblib.dump(model, "resume_model.pkl")
```

```python
# Model evaluation
y_pred = model.predict(X_test_vectorized)

accuracy = accuracy_score(y_test,y_pred)
print("Accuracy:", accuracy)

# testing
sample = ["djangp api backend"]
vec = vectorizer.transform(sample)
predicted_role = model.predict(vec)
print("Predicted Role:", predicted_role[0])
```
